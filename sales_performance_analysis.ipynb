## E-commerce Sales Performance Analysis
We examine a saved monthly product-category export from the synthetic theLook dataset. Run the cells in order from the repository root.


## 1. Load the saved data
Import the shared analysis functions and load the local CSV. The loader checks the columns, missing values, duplicate month-category rows, and metric consistency.


```python
from sales_analysis import load_sales, monthly_summary, category_summary, latest_month_comparison

sales = load_sales()
print(f'Rows: {len(sales):,}; months: {sales.month.nunique()}; categories: {sales.category.nunique()}')
display(sales.head())
```

## 2. Check coverage and definitions
Show the date range and aggregate values. The order count is category order occurrences; the CSV does not contain order IDs for a distinct storewide count.


```python
print('Date range:', sales.month.min().date(), 'to', sales.month.max().date())
print('Revenue:', round(sales.revenue.sum(), 2))
print('Items sold:', int(sales.items_sold.sum()))
print('Category order occurrences:', int(sales.total_orders.sum()))
```

## 3. Review the monthly trend
Aggregate category rows into monthly revenue and item counts. The percentage compares each month with the prior month in this saved snapshot.


```python
monthly = monthly_summary(sales)
display(monthly.tail(12))
ax = monthly.plot(x='month', y='revenue', figsize=(11, 4), legend=False, title='Recorded revenue by month')
ax.set_ylabel('Revenue (source currency)')
```

## 4. Compare product categories
Rank categories by recorded revenue and calculate each category's share. These are descriptive comparisons, not profit or customer value.


```python
categories = category_summary(sales)
display(categories.head(10))
ax = categories.head(10).sort_values('revenue').plot.barh(x='category', y='revenue', figsize=(10, 5), legend=False, title='Top categories by recorded revenue')
ax.set_xlabel('Revenue (source currency)')
```

## 5. Investigate the latest month
Compare category revenue in the latest month with the preceding month. Large differences identify questions for transaction-level follow-up.


```python
changes = latest_month_comparison(sales)
display(changes.head(10))
print('Latest monthly change:', round(monthly.iloc[-1].revenue_change_pct, 1), '%')
```

## 6. Write an evidence-based interpretation
Summarize what the snapshot shows and keep the limitation visible. The last-month jump needs source validation before any causal explanation.


```python
leader = categories.iloc[0]
print(f'Top category: {leader.category}, {leader.revenue:,.2f} ({leader.revenue_share_pct:.1f}% of recorded revenue)')
print('Next check: reconcile the last-month jump against underlying order items, returns, and any data refresh.')
```
